```python

#1. Even or Odd
"""- Get an integer from the user.
- Check whether the number is even or odd using the modulus operator."""
num=int(input("enter the number"))
if num%2==0:
    print("even number")
else:
    print("odd number")
```

```python
#4.Leap Year Checker
"""- Get a year from the user.
 - Check whether it is a leap year or not using conditional statements and logical operators."""
year = int(input("enter the year:"))
if year%400==0 or year%100!=0:
    print("it is a leap year")
else:
    print("it is not a leap year")
           
```

```python

#5.Electricity Bill Calculator
 """Get the number of electricity units consumed and calculate the bill based on the following rates:
- First 100 units: ₹2 per unit
- Next 100 units: ₹3 per unit
 - Above 200 units: ₹5 per unit
 Display the total electricity bill."""
unit=int(input ("enter your electricity units consumed"))
if unit<=100:
    bill=unit*2
elif unit<=200:
    bill=(100*2)+(unit-100)*3
else:
    bill=(100*2)+(100*3)+(unit-200)*5
print("Total electricity bill:",bill)
```

```python

#7.Palindrome Words
"""-Ask the user to enter a string
 -Check whether the string is palindrome or not"""

word=input("enter the word:")
if word==word[::-1]:
    print("palindrome")
else:
    print("not")
```

```python
#9.Traffic Signal
"""- Get a traffic signal colour from the user.
- If it is Red, display "Stop".
 - If it is Yellow, display "Get Ready".
- If it is Green, display "Go".
- Otherwise, display "Invalid Signal"."""

colour=input("enter the colour:")
if colour=="red":
    print("stop")
elif colour=="yellow":
    print("Get ready")
elif colour=="green":
    print("go")
else:
    print("inavalid signal")
             
```

```python
# 10.Simple Calculator
"""- Get two numbers and an operator from the user.
- Perform addition, subtraction, multiplication or division based on the operator.
- Display an appropriate message for an invalid operator."""

num1=int(input("enter the 1st number:"))
num2=int(input("enter the 2nd number:"))
operator=input("enter the operator")
if operator=="+":
    num=num1+num2
elif operator=="-":
    num=num1-num2
elif operator=="*":
    num=num1*num2
elif operator=="/":
    num=num1/num2
else:
    print("invalid operation")
    num=None
if num is not None:
    print("output:",num)
```
